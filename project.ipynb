```python
X = df.drop(columns=['Outcome'])
y = df['Outcome']
X
```

```python
from sklearn.preprocessing import StandardScaler
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
dataset = TensorDataset(X_tensor)
dataloader = DataLoader(dataset, batch_size=64, shuffle=True)
```

```python
class VAE(nn.Module):
    def __init__(self, input_dim, latent_dim=8):
        super(VAE, self).__init__()
        
        # Encoder
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.BatchNorm1d(32),
            nn.ReLU()
        )
        self.mu = nn.Linear(32, latent_dim)
        self.logvar = nn.Linear(32, latent_dim)
        
        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 32),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Linear(32, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, input_dim)
        )

    def encode(self, x):
        h = self.encoder(x)
        return self.mu(h), self.logvar(h)
    
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    
    def decode(self, z):
        return self.decoder(z)
    
    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decode(z)
        return x_recon, mu, logvar
```

```python
vae = VAE(input_dim=X.shape[1], latent_dim=8)
optimizer = optim.Adam(vae.parameters(), lr=0.001)
```

```python
def loss_function(recon_x, x, mu, logvar, beta=0.001):
    recon_loss = nn.functional.mse_loss(recon_x, x, reduction='sum')
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return recon_loss + beta * kl_loss  #
    


vae.train()
for epoch in range(100):
    total_loss = 0
    for batch in dataloader:
        x_batch = batch[0]
        optimizer.zero_grad()
        recon_x, mu, logvar = vae(x_batch)
        loss = loss_function(recon_x, x_batch, mu, logvar)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f"Epoch [{epoch+1}/100], Loss: {total_loss/len(dataloader):.2f}")

```

```python
vae.eval()
with torch.no_grad():
    z = torch.randn(500, 8)  
    synthetic = vae.decode(z).numpy()

synthetic_df = pd.DataFrame(scaler.inverse_transform(synthetic), columns=X.columns)
display(synthetic_df.head())
```

```python
for c in X.columns:
    plt.figure(figsize=(6,3))
    sns.kdeplot(X[c], label='Real', fill=True)
    sns.kdeplot(synthetic_df[c], label='Synthetic', fill=True)
    plt.title(f"Distribution Comparison: {c}")
    plt.legend()
    plt.show()
```

```python
def compare_stats(real, synth):
    real_stats = real.describe().loc[['mean','std']].T
    synth_stats = synth.describe().loc[['mean','std']].T
    diff = (synth_stats - real_stats).abs()
    return pd.concat([real_stats, synth_stats, diff], axis=1, keys=['Real','Synthetic','|Diff|'])

stats = compare_stats(X, synthetic_df)
display(stats.round(2))
```
